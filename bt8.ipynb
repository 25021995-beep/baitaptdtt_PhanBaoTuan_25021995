```python
class cylinder():
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height
    def circumference(self):
        return 2*3.14*self.radius * self.height+2*(self.radius)*(self.radius)*3.14
    def volume(self):
        return 3.14*self.radius*self.radius * self.height
a=list(map(float,input().split()))
b=cylinder(a[0],a[1])
print(b.circumference())
print(b.volume())
```

```python
class date():
    def __init__(self,day,month,year):
        self.year = year
        self.month = month
        self.day = day
    def valid(self):
        if self.year < 0 or self.month<0 or self.month>12 or self.day >31:
            return "Invalid date"
        if self.month in (4,6,9,11) and self.day>30:
            return "Invalid date"
        if self.year%4==0 and self.year%400!=0 and self.month==2 and self.day>29:
            return "Invalid date"
        if self.year%4!=0 and self.month==2 and self.day>28:
            return "Invalid date"

    def display(self):
        if self.valid()!="Invalid date":
            if self.year%4==0 and self.year%400!=0 and self.month==2 and self.day==29:
                self.day=1
                self.month=3
                print(f"{self.day}//{self.month}//{self.year}")
            elif self.year%4!=0 and self.month==2 and self.day==28:
                self.day=1
                self.month=3
                print(f"{self.day}//{self.month}//{self.year}")
            elif self.month in (4,6,9,11) and self.day==30:
                self.day=1
                print(f"{self.day}//{self.month+1}//{self.year}")

            elif self.month==12 and self.day==31:
                self.month=1
                self.year=self.year+1
                print(f"{self.day}//{self.month}//{self.year}")
            else:
                print(f"{self.day+1}//{self.month}//{self.year}")
        elif self.valid()=="Invalid date":
            return print("Invalid date")

a=list(map(str,input().split("//")))
a=list(map(int,a))
date=date(a[0],a[1],a[2])

date.display()
```

```python
import math


class point():
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def isOriginal(self):
        if self.x==0 and self.y==0:
            return True
        return False
    def isHorizontal(self):
        if self.y==0:
            return True
        return False
    def isVertical(self):
        if self.x==0:
            return True
        return False
    def distance(self):
        dist= self.x**2 + self.y**2
        dist = dist**0.5
        dist=round(dist,2)
        return dist
a=list(map(int,input().split()))
if point(a[0],a[1]).isOriginal()==True:
    print("point is ORIGINAL")
elif point(a[0],a[1]).isHorizontal()==True:
    print("point is HORIZAL")
elif point(a[0],a[1]).isVertical()==True:
    print("point is VERTICAL")
print(point(a[0],a[1]).distance())
```

```python
class Rectangle():
    def __init__(self,w,h,k):
        self.w=k*w
        self.h=k*h
    def area(self):
        return round(self.w*self.h,2)
    def perimeter(self):
        return round(2*(self.w+self.h),2)
a=list(map(float,input(" w h k:").split()))
print(Rectangle(a[0],a[1],a[2]).area(),Rectangle(a[0],a[1],a[2]).perimeter())
```

```python
class Fraction():
    def __init__(self,num,den):
        self.num=num
        self.den=den
    def shortcut(self):
        a=[]
        b=[]
        c=[]
        if self.num>0:
            for i in range(1,self.num+1):
                if self.num%i==0:
                    a.append(i)
            for j in range(1,self.den+1):
                if self.den%j==0:
                    b.append(j)

            for i in a:
                if i in b:
                    c.append(i)
            return (f"{int(self.num/max(c))}/{int(self.den/max(c))}")
        elif self.num<0:
            self.num=-self.num
            if self.num > 0:
                for i in range(1, self.num+1):
                    if self.num % i == 0:
                        a.append(i)
                for j in range(1, self.den+1):
                    if self.den % j == 0:
                        b.append(j)

                for i in a:
                    if i in b:
                        c.append(i)
                return (f"{-int(self.num / max(c))}/{int(self.den / max(c))}")





a=list(map(str,input().split()))
f1=Fraction(int(a[0]),int(a[1]))
f2=Fraction(int(a[3]),int(a[4]))
op=a[2]
if op=="+":
    tu=f1.num*f2.den+f1.den*f2.num
    mau=f1.den*f2.den
    print(Fraction(tu,mau).shortcut())
if op=="-":
    tu=f1.num*f2.den-f1.den*f2.num
    mau=f1.den*f2.den
    print(Fraction(tu,mau).shortcut())
if op=="*":
    tu=f1.num*f2.num
    mau=f1.den*f2.den
    print(Fraction(tu,mau).shortcut())
if op=="/":
    tu=f1.num*f2.den
    mau=f1.den*f2.num
    print(Fraction(tu,mau).shortcut())






```

```python
class BankAccount():
    def __init__(self,onwer,balance):
        self.onwer=onwer
        self.balance=float(balance)
    def deposit(self,amount):
        self.balance+=float(amount)
    def withdraw(self,amount):
        if self.balance-float(amount)<0:
            self.balance=self.balance
        elif self.balance-float(amount)>=0:
            self.balance-=float(amount)
    def __str__(self):
        return round(self.balance,2)
a=list(map(str,input().split()))
account=BankAccount(a[0],float(a[1]))
n=int(input())
for i in range(n):
    b=list(map(str,input().split()))
    if b[0]=='WITHDRAW':
        account.withdraw(float(b[1]))
    elif b[0]=='DEPOSIT':
        account.deposit(float(b[1]))
print(account.__str__())
```

```python
class Student():
    def __init__(self,name):
        self.subject = None
        self.score= None
        self.name=name
    def add_score(self,subject,score):
        self.subject=subject
        self.score=score
    def avg_score(self):
        sum=0
        for i in self.score:
            sum+=float(i)
        sum=sum/len(self.score)
        return float(sum)
    def rank(self):
        if self.avg_score()>=8:
            return "Excellent"
        elif 6.5<=self.avg_score()<8:
            return "Good"
        elif 5<=self.avg_score()<6.5:
            return "Average"
        elif 0<=self.avg_score()<5:
            return "Poor"
    def __str__(self):
        return f'{self.name} {round(self.avg_score(),2)} {self.rank()}'


name=input()
n=int(input())
subject=[]
score=[]
for i in range(n):
    c=list(map(str,input().split()))
    subject.append(c[0])
    score.append(float(c[1]))
student=Student(name)
student.add_score(subject,score)
print(student.__str__())
```

```python
import math
class point():
    def __init__(self,x,y,z):
        self.x = x
        self.y = y
        self.z = z
p1=list(map(float,input().split()))
p2=list(map(float,input().split()))
p3=list(map(float,input().split()))
p4=list(map(float,input().split()))
A=point(p1[0],p1[1],p1[2])
B=point(p2[0],p2[1],p2[2])
C=point(p3[0],p3[1],p3[2])
D=point(p4[0],p4[1],p4[2])
AB=point(B.x-A.x,B.y-A.y,B.z-A.z)
AC=point(C.x-A.x,C.y-A.y,C.z-A.z)
BC=point(C.x-B.x,C.y-B.y,C.z-B.z)
BD=point(D.x-B.x,D.y-B.y,D.z-B.z)
ABxAC=point(AB.y*AC.z-AC.y*AB.z,AB.z*AC.x-AC.z*AB.x,AB.x*AC.y-AC.x*AB.y)
BCxBD=point(BC.y*BD.z-BD.y*BC.z,BC.z*BD.x-BD.z*BC.x,BC.x*BD.y-BD.x*BC.y)
length_ABxAC=((ABxAC.x)**2+(ABxAC.y)**2+(ABxAC.z)**2)**0.5
length_BCxBD=((BCxBD.x)**2+(BCxBD.y)**2+(BCxBD.z)**2)**0.5

cosine=(ABxAC.x*BCxBD.x+ABxAC.y*BCxBD.y+ABxAC.z*BCxBD.z)/(length_ABxAC*length_BCxBD)
theta=math.acos(cosine)
theta=theta*180/math.pi
print(round(theta,2))

```

```python
class Complex():
    def __init__(self,real,imag):
        self.real=real
        self.imag=imag
def cal(a,b):
    real=a.real*b.real-a.imag*b.imag
    imag=a.real*b.imag+b.real*a.imag
    print(f"{real} {imag}")

a=list(map(int,input().split()))
com1=Complex(a[0],a[1])
b=list(map(int,input().split()))
com2=Complex(b[0],b[1])
cal(com1,com2)

```

```python
class triagle():
    def __init__(self,edge1,edge2,edge3):
        self.edge1=int(edge1)
        self.edge2=int(edge2)
        self.edge3=int(edge3)
def area(segment):
    if segment.edge1+segment.edge2<=segment.edge3 or segment.edge2+segment.edge3<=segment.edge1 or segment.edge3+segment.edge1<=segment.edge2:
        return "Invalid"
    else:
        p=(segment.edge1+segment.edge2+segment.edge3)/2
        S=(p*(p-segment.edge1)*(p-segment.edge2)*(p-segment.edge3))**0.5
        return round(S,2)
a=list(map(int,input().split()))
tri1=triagle(a[0],a[1],a[2])
print(area(tri1))

```
